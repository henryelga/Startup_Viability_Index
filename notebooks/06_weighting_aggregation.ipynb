# 6. Weighting and Aggregation

## 6.1. Introduction

After normalisation, the indicators are combined into sub-index scores and an overall composite index. 

This stage requires selecting appropriate weighting and aggregation methods that are consistent with the theoretical structure of the framework.

### Weighting Approach

Equal weighting is applied across all indicators within each sub-index. 

This assumes that each indicator contributes equally to its dimension and avoids introducing subjective weights without strong theoretical justification.

### Aggregation Method

Arithmetic mean aggregation is used to combine indicators into sub-index scores and to calculate the final composite index. 

This approach is simple, transparent, and widely used in composite indicator construction.

### Correlation and Redundancy Considerations

Potential correlation and redundancy between indicators were examined earlier using correlation analysis, Variance Inflation Factor (VIF) testing, and Principal Component Analysis (PCA). 

Based on these results, the highly redundant variable `stock_market_liquidity` was removed before the aggregation stage.

### Compensability Between Indicators

The use of arithmetic mean aggregation allows compensability between indicators. 

This means that weaker performance in one area can be offset by stronger performance in another area within the composite index framework.


```python
import os
import pandas as pd

normalised_data_path = "../data/worldbank_normalised_data.csv"
```

```python
df_index = pd.read_csv(normalised_data_path)

# Sub-index groupings
sub_indices = {

    "Finance_Investment": [
        "credit_to_private_sector",
        "fdi_inflows",
        "market_capitalisation"
    ],

    "Technology_Innovation": [
        "broadband_subscriptions",
        "internet_usage_rate",
        "secure_servers_density",
        "rd_expenditure"
    ],

    "Economic_Growth_Stability": [
        "gdp_growth",
        "inflation_rate",
        "real_interest_rate",
        "capital_formation"
    ],

    "Talent_Productivity": [
        "tertiary_enrollment",
        "labor_productivity",
        "high_tech_exports",
        "ict_imports"
    ]
}
```

## 6.2. Sub-Index Scores Construction

Sub-index scores are calculated by taking the arithmetic mean of the normalised indicators within each dimension. Since equal weighting is applied, each indicator contributes equally to its corresponding sub-index score.

```python
# Calculate sub-index scores
for subindex_name, indicators in sub_indices.items():

    df_index[subindex_name] = df_index[indicators].mean(axis=1)

print("=========================================================================================================================")
print("                                                Sub-Index Scores Calculated                                              ")
print("=========================================================================================================================")

print(
    df_index[
        [
            "Country Name",
            "Finance_Investment",
            "Technology_Innovation",
            "Economic_Growth_Stability",
            "Talent_Productivity"
        ]
    ].head(20).to_string(index=False)
)
```

These sub-index scores provide a summary measure for each major dimension of the framework. Each score reflects a country’s relative performance within a specific area after combining the corresponding normalised indicators.

These sub-index values are then used to calculate the final composite index in the next stage.

## 6.3. Composite Index (Startup Viability Index) Construction

The final Startup Viability Index is calculated by taking the arithmetic mean of the four sub-index scores. Since equal weighting is applied across all dimensions, each sub-index contributes equally to the final composite index value.

```python
# Calculate final Startup Viability Index

subindex_columns = [
    "Finance_Investment",
    "Technology_Innovation",
    "Economic_Growth_Stability",
    "Talent_Productivity"
]

df_index["Startup_Viability_Index"] = df_index[subindex_columns].mean(axis=1)

print("========================================================")
print("            Startup Viability Index Calculated          ")
print("========================================================")

print(
    df_index[
        [
            "Country Name",
            "Startup_Viability_Index"
        ]
    ].head(20).to_string(index=False)
)
```

The Startup Viability Index combines the four sub-index dimensions into a single overall score for each country. Higher values indicate stronger overall performance across the framework, while lower values indicate weaker overall performance relative to other countries.

The composite index scores are used in the next stage to rank countries and compare overall performance across the dataset.

## 6.4. Export of Final Index Results

After calculating the sub-index scores and the final Startup Viability Index (SVI), the completed dataset is exported for use in the ranking and interpretation stages of the analysis.

```python
output_results_dir = "../data/results"
os.makedirs(output_results_dir, exist_ok=True)

# Export sub-index scores
df_index[
    [
        "Country Name",
        "Country Code",
        "Income Group",
        "Finance_Investment",
        "Technology_Innovation",
        "Economic_Growth_Stability",
        "Talent_Productivity"
    ]
].to_csv(
    os.path.join(output_results_dir, "subindex_scores.csv"),
    index=False
)

# Export final SVI results
df_index[
    [
        "Country Name",
        "Country Code",
        "Income Group",
        "Startup_Viability_Index"
    ]
].to_csv(
    os.path.join(output_results_dir, "startup_viability_index.csv"),
    index=False
)

print("Results Exported Successfully")
print(f"Saved to: {output_results_dir}")
```

## 6.5. Country Rankings

Countries are ranked based on their final Startup Viability Index (SVI) scores. Higher SVI values indicate stronger overall performance across the four dimensions of the framework, while lower scores indicate weaker overall startup ecosystem conditions.

The rankings provide a comparative view of how countries perform relative to one another across finance, technology, economic stability, and talent indicators.

### 6.5.1. Top 20 Countries

```python
# Rank countries by SVI score

df_rankings = df_index.sort_values(
    by="Startup_Viability_Index",
    ascending=False
).reset_index(drop=True)

df_rankings.insert(0, "Rank", range(1, len(df_rankings) + 1))

print("======================================================")
print("                 Top 20 Countries by SVI")
print("======================================================")

print(
    df_rankings[
        [
            "Rank",
            "Country Name",
            "Startup_Viability_Index"
        ]
    ].head(20).to_string(index=False)
)
```

The top 20 countries in the Startup Viability Index (SVI) show strong overall performance across all four sub-indices. 

These countries typically perform well across all four dimensions, particularly in financial development, technological infrastructure, and human capital. 

Their high scores reflect more advanced economic structures, stronger institutional capacity, and greater levels of innovation and productivity, which together support a more favourable environment for startups.

### 6.5.2. Bottom 20 Countries

```python
print("========================================================")
print("               Bottom 20 Countries by SVI")
print("========================================================")

print(
    df_rankings[
        [
            "Rank",
            "Country Name",
            "Startup_Viability_Index"
        ]
    ].tail(20).to_string(index=False)
)
```

The bottom 20 countries in the SVI generally have weaker performance across most indicators. 

These countries generally face constraints such as limited financial market development, lower digital infrastructure, and reduced investment in education and innovation. 

As a result, their overall startup ecosystem conditions are less developed, leading to consistently lower SVI scores across all sub-indices.
