# 1. Theoretical Framework: Startup Viability Index (SVI)

## 1.1. Introduction and Project Context
The aim of this project is to create a **Startup Viability Index (SVI)** that measures how supportive different countries are for startups and entrepreneurship.

I chose this topic because of my own interest in startups and innovation. During our final year project, my team developed a project called **Stockd**, which was selected as:

- the Best Team Project at the Expo 2026
- and also shortlisted in the Top 50 Student Entrepreneur Awards

As a team, we started discussing the possibility of continuing the project as a real startup after university. This made me curious about why some countries are better places for startups than others. I wanted to understand:

- what makes a country startup-friendly
- what conditions help startups grow
- and how these conditions can be measured using data.

To answer these questions, I conducted extensive research to build a data-driven tool: the **Startup Viability Index (SVI)**. This framework outlines the reasons behind the metrics I chose to predict startup viability.

## 1.2. Problem Statement and Objectives

Starting a new business is risky, and most startups fail within their first few years. Many existing economic tools only look at basic metrics like country GDP or simple business registration steps. However, a modern startup needs a complex mix of funding, technology, stable prices, and smart people to survive.

The main goal of the **SVI** index is to measure how friendly an environment is for a startup throughout its lifecycle. By combining 16 data variables into 4 distinct sub-indices, my index creates a clear score to evaluate a country's startup environment.

## 1.4. Sub-Indices and Variables Justification

In line with the OECD Handbook’s requirement for analytical soundness, I used expert opinions and academic research to justify why these specific variables determine startup success:

### Sub-Index 1: Finance & Investment

This sub-index measures how easily money flows into new businesses. Startups need different ways to get cash: like bank loans, global investors, or stock markets, to survive the early days when they aren't making a profit yet.

* **Research Justification:** I based this on research by [Trojette (2023)](https://ecommons.luc.edu/meea/1332/). This study shows that strong local banks (credit) and active stock markets (market cap) are the main reasons why foreign investors (FDI) bring their money into a country. A deep financial market gives startups the cash runway they need to grow.

* **Variables I am using to measure this:**
  * *Domestic credit to private sector (% of GDP):* Shows how easy it is to get traditional bank loans.
  * *Foreign direct investment, net inflows (% of GDP):* Measures how much international cash is coming into the country.
  * *Market capitalization of listed domestic companies (% of GDP):* Shows the overall size and health of the local stock market.
  * *Stocks traded, total value (% of GDP):* Measures how active and liquid the stock market is.

### Sub-Index 2: Technology & Innovation

In today's world, good tech is not optional. Startups rely heavily on digital tools to find customers, run operations, and keep costs low. 

* **Research Justification:** I found support for this in a study by [Leogrande et al. (2022)](https://mpra.ub.uni-muenchen.de/112246/). Their research explicitly links fast internet (broadband) and safe web infrastructure (secure servers) to better company performance and a higher chance of attracting venture capital. Also, high R&D spending means the country is actively creating the new technologies that tech startups commercialize.

* **Variables I am using to measure this:**
  * *Fixed broadband subscriptions (per 100 people):* Measures the physical, high-speed internet infrastructure.
  * *Individuals using the Internet (% of population):* Shows how big the digital customer base is.
  * *Secure Internet servers (per 1 million people):* Measures how safe and mature the country is for e-commerce.
  * *Research and development (R&D) expenditure (% of GDP):* Shows the country’s commitment to creating new tech.

### Sub-Index 3: Economic Growth & Stability

This sub-index looks at the overall health of the economy. A startup can have the best product in the world, but if the local economy is unstable, has massive inflation, or has crazy high interest rates, the business will likely fail.

* **Research Justification:** This selection is backed by [Arin et al. (2013)](https://journals.sagepub.com/doi/10.1177/0149206312455838). Using advanced statistical models, they proved that stable GDP growth and low inflation are **universal** predictors of how many people start businesses. Steady growth means consumers have money to spend, and stable interest rates keep the cost of borrowing manageable for young startups.

* **Variables I am using to measure this:**
  * *GDP growth (annual %):* Shows if the overall economy is expanding and creating demand.
  * *Inflation, consumer prices (annual %):* Measures price stability (high inflation kills profit margins).
  * *Real interest rate (%):* Shows the true cost of borrowing operational capital.
  * *Gross fixed capital formation (% of GDP):* Measures investments in physical things like roads and machinery.

### Sub-Index 4: Talent & Productivity

A startup ecosystem needs more than just money and cables; it needs brainpower. This sub-index measures how smart, productive, and tech-savvy the local workforce is.

* **Research Justification:** This is justified by [Ratzinger et al. (2017)](https://link.springer.com/article/10.1007/s10961-017-9627-3). Their empirical research proves that higher education (tertiary enrollment) is the single strongest predictor of whether a startup will hit its funding and growth milestones. Advanced education directly fuels high-tech exports and labor productivity, giving startups a highly skilled talent pool to hire from.

* **Variables I am using to measure this:**
  * *School enrollment, tertiary (% gross):* Shows the availability of highly educated, university-level talent.
  * *GDP per person employed (constant 2021 PPP $):* Measures the baseline productivity and efficiency of workers.
  * *High-technology exports (% of manufactured exports):* Reflects how competitive the country is at making complex tech.
  * *ICT goods imports (% total goods imports):* Shows how well the country can access global tech parts and hardware.

## 1.6. Methodology: Choice of Composite Index Approach

Using a composite index approach is the best way to solve this problem. Because the data includes completely different units (like percentages, growth rates, and counts per capita), compiling them into one unified index allows us to normalize the data. 

As outlined in the *Handbook on Constructing Composite Indicators*, the following steps will address data normalization to make these variables comparable, ensuring the final index is mathematically sound and suitable for assessing whether a country provides a favourable environment for startup growth and survival.

## 1.7. Data Sourcing and Handbook Criteria

All data used to construct this index was retrieved from the [World Bank DataBank](https://databank.worldbank.org/). Choosing the World Bank explicitly satisfies the **Step 2 (Data Selection)** criteria from the OECD Handbook Checklist:

Country Coverage: The World Bank provides large, standardised datasets across many countries, which reduces the need for missing data handling later (Step 3).
Data Quality: The data is collected from official national sources and international organisations, making it reliable and consistent.
Comparability: The World Bank uses the same methods across countries, which allows fair comparisons between them and avoids inconsistent measurements. This makes the data suitable for the later analysis and normalisation steps.
